# What is Dependency Parsing?

Dependency parsing is a method for analyzing the **syntactic structure of a sentence**.

Instead of grouping words into phrases, dependency parsing represents a sentence as a **set of relations between words**.

Each relation connects:

- a **head** (central word),
- a **dependent** (word that modifies or complements the head).

This forms a **tree structure**.

---

## Key idea

Each word (token) in a sentence:

- has a **dependency label** (`dep_`)
- depends on another word (its **head**)

---

## Example

Sentence:
> "The cat eats fish."

Dependencies:

- "cat" → subject of "eats"
- "fish" → object of "eats"
- "eats" → ROOT (main verb)

---

## In spaCy

For each token:

- `token.dep_` → dependency label
- `token.head` → head word
- `token.children` → words depending on this token

## Download and load the Treebank corpus

NLTK corpora are usually downloaded separately from the Python package itself. According to the NLTK documentation, corpora can be installed through the NLTK downloader and then accessed through `nltk.corpus`.
The Treebank corpus is useful here because it contains well-formed English sentences.

```python
# Download the Treebank corpus from NLTK.
nltk.download("treebank")

# Import the corpus reader.
from nltk.corpus import treebank
```

```python
# Load the sentences from the corpus.
sentences = treebank.sents()
```

```python
# Convert the first sentence from a list of tokens into one text string.
first_sentence_text = " ".join(sentences[0])

# Print the result.
print(first_sentence_text)
```

```python
# Go through the first 5 sentences in the corpus.
texts = []
for sentence in sentences[:5]:
    text = " ".join(sentence)
    texts.append(text)
```

```python
# Print the prepared texts.
for text in texts:
    print(text)
    print()
```

```python
# Process the text with spaCy.
nlp = spacy.load("en_core_web_sm")
text = texts[0]
doc = nlp(text)
```

```python
for token in doc:
    print(token.text)
```

```python
for token in doc:# doc for spacy
    print("TOKEN:", token.text)
    print("DEPENDENCY:", token.dep_) #dep_ has to be the thing
    print("HEAD:", token.head.text) #
    print("---") #just for the visuals lol
```

# Understanding Sentence Structure

We now focus on the structure of a parsed sentence.

We will move step by step:

1. ROOT (main verb)
2. children (direct dependents)
3. subject
4. object
5. full structure
6. subtree (phrases)
7. visualization

We always work on ONE sentence at a time. _> (not a full text)

**Find the ROOT of the sentence**

The `ROOT` is the central word in the sentence structure.
It is often the main verb.

```python
for token in doc:
    if token.dep_ == "ROOT":
        print("The ROOT of the sentence is:", token.text)
```

```python
# Exercise 1. Find the subject. The subject is usually marked as "nsubj".
for token in doc:
    if token.dep_ == "nsubj":# dependency label
      print("The subject is:", token.text)

       # finish the code
```

## Children of a token

Each token can have **children**.

Children are words that directly depend on a given word.

For example:
- subjects, objects, modifiers are often children of the ROOT.

```python
for sentence_obj in doc.sents:
    # 'sentence_obj' is a spaCy Span object representing a sentence.
    # Get the root token of this sentence.
    # The 'root' attribute of a Span object returns its root token.
    root_token = sentence_obj.root

    print(f"Children of the ROOT ('{root_token.text}') for sentence:")
    # Iterate over the children (dependent tokens) of the root token.
    for child_token in root_token.children:
        print(child_token.text, "|", child_token.dep_)
```

```python
# Find the object
for token in doc:
    if token.dep_ in ["dobj", "obj"]: #it's where you change the label
        print("Object:", token.text)
        print("Verb:", token.head.text) # root part (head)
```

## Extract a phrase

A subtree contains a word and all words connected to it.

```python
for token in doc: #finish the code
    if token.dep_ == "nsubj":
        words = []

        for t in token.subtree:
            words.append(t.text)

        print("Subject phrase:", " ".join(words))
```

## Visualize and save the dependency tree

We can display the dependency tree inside the notebook.

## Save the visualization

`displacy.render()` can return HTML if we do not display it directly.

We can then save that HTML to a file.

Complete the code below.

```python
# Step 1: generate HTML
html = displacy.render(doc, style="dep", jupyter=False, page=True)

# Step 2: open a file
file = open("dependency_tree.html", "w")

# Step 3: write HTML to file
file.write(html)

# Step 4: close file
file.close()
```
