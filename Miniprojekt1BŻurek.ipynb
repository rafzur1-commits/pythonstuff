```python
import pandas as pd

df = pd.read_csv("/content/Fake.csv") # reading the csv
display(df.head()) #
```

```python
filename = "Fake.csv"   #inspecting the columns
df.head()
df.columns
```

```python
df.columns # Minimally cleaning the title column
text_col = df['title']
df = df.dropna()
df = df.drop_duplicates()
df['title'] = df['title'].str.strip().str.lower()
```

```python
text_col = df['title'] #showing the cleaned up column

print(text_col.head())
```

# **Exploring**

```python
df['length'] = df['title'].apply(len) #getting statistics

print(df['length'].describe())

from collections import Counter

all_words = " ".join(df['title']).split()
word_freq = Counter(all_words)

print(word_freq.most_common(15))
```

# **Picking NER for a spacy feature**

```python
import spacy
nlp = spacy.load("en_core_web_sm")
```

```python
docs = list(nlp.pipe(df['title'])) #applying spaCy to the title column
```

```python
entities = [] #showing what the entities are classified as

for doc in docs:
    for ent in doc.ents:
        entities.append((ent.text, ent.label_))

entities[:20]
```

```python
for doc in docs[:5]: #showing it in example
    print("TEXT:", doc.text)
    print("ENTITIES:", [(ent.text, ent.label_) for ent in doc.ents])
    print("-" * 40)
```
